## Funktion `generate_instance` – Instanzgenerator für Aufgabe 2 (Optimierung)

Die Funktion `generate_instance(...)` erzeugt zufällige Testinstanzen für die in **Aufgabe 2** beschriebene Planungsaufgabe der Mensa-Küche.

Ziel ist es, realistische Daten für folgende Elemente zu generieren:

- Menge der Gerichte \( I \)
- Menge der Bearbeitungsphasen \( P \)
- Verfügbare Workstations pro Phase
- Bearbeitungszeiten je Gericht und Phase
- Optional: Soll-Fertigstellungszeiten (Auslagezeitpunkte) \( d_i \)

Diese Daten können anschließend zur Modellierung und Lösung des Optimierungsproblems (Minimierung der Gesamtverspätung) verwendet werden.

---
### Input (Parameter):

- n : Anzahl der Gerichte
- K : Anzahl der Bearbeitungsphasen (immer 5)
- max_machines : Maximale Anzahl paralleler Workstations pro Phase
- p_min, p_max : Unter- und Obergrenze für die Bearbeitungszeiten
- machines_min : Minimale Anzahl paralleler Workstations pro Phase
- due_date_factor : Faktor zur Erzeugung der Soll-Fertigstellungszeiten
- generate_due_dates : Boolescher Wert (True/False) – ob Soll-Fertigstellungszeiten erzeugt werden
- seed : Optionaler Startwert für den Zufallszahlengenerator (für Reproduzierbarkeit)

---

### Verarbeitung / Funktionsweise:

#### Gerichte:

Es wird eine Menge 
I={0,1,…,n−1}
I={0,1,…,n−1} erstellt.

#### Bearbeitungsphasen:

Phasen 
P={0,1,…,K−1}
P={0,1,…,K−1} entsprechen den Prozessschritten:

1. Frische Bestandteile schneiden
2. Beilage vorkochen
3. Frittieren
4. Soße anrühren
5. Gericht final garen


#### Workstations:

Für jede Phase wird eine zufällige Anzahl paralleler Workstations erzeugt.


#### Bearbeitungszeiten:

Für jedes Gericht und jede Phase wird eine ganzzahlige Bearbeitungszeit p[i,k] erzeugt.

Modelliert die Zeit, die ein Gericht auf der Workstation benötigt.


#### Soll-Fertigstellungszeiten (optional):

Falls aktiviert, wird d[i] berechnet: Summe der Bearbeitungszeiten eines Gerichts multipliziert mit einem Zufallsfaktor zwischen 1.0 und due_date_factor.

Sonst d = None.

---

### Output (Rückgabewerte):

- I : Liste der Gerichte
- P : Liste der Bearbeitungsphasen
- M : Dictionary der Workstations je Phase
- p : Dictionary der Bearbeitungszeiten je Gericht und Phase
- d : Dictionary der Soll-Fertigstellungszeiten (Auslagezeitpunkte) oder None

```python
import random

def generate_instance(n, K=5, max_machines,
                      p_min=1, p_max=20,
                      machines_min=2,
                      due_date_factor=1.3,
                      generate_due_dates=True,
                      seed=None):

    if seed is not None:
        random.seed(seed)

    # -------------------
    # Sets
    # -------------------

    I = list(range(n))
    P = list(range(K))

    # Workstations per phase (random)
    M = {}
    for k in P:
        num_machines = random.randint(machines_min, max_machines)
        M[k] = list(range(num_machines))

    # -------------------
    # Processing times p[i,k]
    # -------------------

    p = {}
    for i in I:
        for k in P:
            if random.random() < 0:
                p[i, k] = 0
            else:
                p[i, k] = random.randint(p_min, p_max)

    # -------------------
    # Due dates (optional)
    # -------------------

    d = None

    if generate_due_dates:
        d = {}
        for i in I:
            total_processing = sum(p[i, k] for k in P)
            d[i] = int(total_processing * random.uniform(1.0, due_date_factor))

    return I, P, M, p, d
```

## Eine Beispiel Instanz:

```python
n = 30
K = 5
max_machines = 6


I, P, M, p, d = generate_instance(n, K, max_machines,generate_due_dates=True, seed = 2)
```
